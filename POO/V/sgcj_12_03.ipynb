```python
lista_jogadores = []
# lista_jogadores = list() # segunda forma de criação de listas em python

# Cadastrar novos jogadores
prefixo_id = 202511900001
for i in range(3):
  nome = input("Informe o nome do " + str(i+1) + "º jogador: ")
  senha = input("Informe uma senha para " + nome + ":")
  jogador = {"id":prefixo_id + i, "user":nome, "password":senha, "pontuacao":0}
  lista_jogadores.append(jogador) # adiciona o jogador na lista
```
